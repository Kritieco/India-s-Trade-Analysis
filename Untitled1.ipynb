```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression
import os

folder_path = r"C:\Users\kriti\Downloads\trade_project"

# ── STEP 1: LOAD ──────────────────────────────────────────────
df = pd.read_csv(os.path.join(folder_path, "2010_2021_HS2_export.csv"), encoding="latin1")

# ── STEP 2: CLEAN ─────────────────────────────────────────────
df.columns = df.columns.str.strip().str.lower()
df = df.rename(columns={"hscode":"HS_Code","commodity":"Commodity","value":"Export_Value","country":"Country","year":"Year"})
df = df.dropna(subset=["Export_Value"])
df["Export_Value"] = pd.to_numeric(df["Export_Value"], errors="coerce")
df = df[df["Export_Value"] > 0]
df["Country"]   = df["Country"].str.strip().str.upper()
df["Commodity"] = df["Commodity"].str.strip().str.title()
print(f"✓ Clean data: {df.shape[0]:,} rows")

# ── STEP 3: YEARLY TREND ──────────────────────────────────────
print("\n--- Total Exports by Year ---")
yearly = df.groupby("Year")["Export_Value"].sum().reset_index()
yearly.columns = ["Year","Total_Exports"]
yearly["YoY_Growth_%"] = yearly["Total_Exports"].pct_change().mul(100).round(2)
print(yearly.to_string(index=False))

# ── STEP 4: TOP 10 COUNTRIES ──────────────────────────────────
print("\n--- Top 10 Export Partners ---")
top_countries = (df.groupby("Country")["Export_Value"]
                 .sum().sort_values(ascending=False).head(10).reset_index())
top_countries.columns = ["Country","Total_Exports"]
print(top_countries.to_string(index=False))

# ── STEP 5: TOP 10 COMMODITIES ────────────────────────────────
print("\n--- Top 10 Exported Commodities ---")
top_commodities = (df.groupby("Commodity")["Export_Value"]
                   .sum().sort_values(ascending=False).head(10).reset_index())
top_commodities.columns = ["Commodity","Total_Exports"]
top_commodities["Commodity"] = top_commodities["Commodity"].str[:45]
print(top_commodities.to_string(index=False))

# ── STEP 6: CAGR BY COUNTRY ───────────────────────────────────
print("\n--- CAGR by Country (2010-2021) ---")
cagr_results = []
for country in df["Country"].unique():
    temp = df[df["Country"]==country].groupby("Year")["Export_Value"].sum()
    if 2010 in temp.index and 2021 in temp.index:
        start, end = temp[2010], temp[2021]
        if start > 0:
            cagr = ((end/start)**(1/11)-1)*100
            cagr_results.append([country, round(cagr,2)])
cagr_df = pd.DataFrame(cagr_results, columns=["Country","CAGR_%"]).sort_values("CAGR_%", ascending=False)
print("Top 10 fastest growing:")
print(cagr_df.head(10).to_string(index=False))

# ── STEP 7: COUNTRY SHARE ─────────────────────────────────────
country_share = (df.groupby("Country")["Export_Value"].sum() /
                 df["Export_Value"].sum() * 100).sort_values(ascending=False)

# ── STEP 8: REGRESSION FORECAST ──────────────────────────────
print("\n--- Export Forecast ---")
X = yearly[["Year"]]
y = yearly["Total_Exports"]
model = LinearRegression()
model.fit(X, y)
for yr in [2022, 2023, 2024]:
    pred = model.predict(pd.DataFrame([[yr]], columns=["Year"]))[0]
    print(f"  Predicted exports {yr}: {pred:,.2f} USD Million")

# ── STEP 9: SAVE CLEAN FILE ───────────────────────────────────
out = os.path.join(folder_path, "india_exports_clean.xlsx")
with pd.ExcelWriter(out, engine="openpyxl") as w:
    yearly.to_excel(w, sheet_name="Yearly Trend", index=False)
    top_countries.to_excel(w, sheet_name="Top Countries", index=False)
    top_commodities.to_excel(w, sheet_name="Top Commodities", index=False)
    cagr_df.to_excel(w, sheet_name="CAGR by Country", index=False)
print(f"\n✓ Clean file saved: {out}")

# ── STEP 10: CHARTS ───────────────────────────────────────────
fig, axes = plt.subplots(2, 3, figsize=(18,10))
fig.suptitle("India Export Analysis (2010–2021)", fontsize=15, fontweight="bold")

# Chart 1 — Yearly trend
ax1 = axes[0,0]
ax1.plot(yearly["Year"], yearly["Total_Exports"], marker="o", color="#1B6B5C", linewidth=2)
ax1.set_title("Total Exports by Year")
ax1.set_ylabel("Export Value (USD Mn)")
ax1.yaxis.set_major_formatter(plt.FuncFormatter(lambda x,_: f"{x:,.0f}"))

# Chart 2 — Top countries
ax2 = axes[0,1]
tc = top_countries.sort_values("Total_Exports")
ax2.barh(tc["Country"], tc["Total_Exports"], color="#1B6B5C")
ax2.set_title("Top 10 Export Partners")
ax2.xaxis.set_major_formatter(plt.FuncFormatter(lambda x,_: f"{x:,.0f}"))

# Chart 3 — Top commodities
ax3 = axes[0,2]
tcm = top_commodities.sort_values("Total_Exports")
ax3.barh(tcm["Commodity"].str[:30], tcm["Total_Exports"], color="#D85A30")
ax3.set_title("Top 10 Commodities")
ax3.xaxis.set_major_formatter(plt.FuncFormatter(lambda x,_: f"{x:,.0f}"))

# Chart 4 — Country share pie
ax4 = axes[1,0]
top8 = country_share.head(8)
others = pd.Series({"OTHERS": country_share.iloc[8:].sum()})
pie_data = pd.concat([top8, others])
ax4.pie(pie_data, labels=pie_data.index, autopct="%1.1f%%",
        startangle=140, colors=plt.cm.Set3.colors)
ax4.set_title("Export Share by Country")

# Chart 5 — CAGR
ax5 = axes[1,1]
top_cagr = cagr_df.head(10).sort_values("CAGR_%")
colors = ["#1D9E75" if x > 0 else "#D85A30" for x in top_cagr["CAGR_%"]]
ax5.barh(top_cagr["Country"], top_cagr["CAGR_%"], color=colors)
ax5.axvline(0, color="black", linewidth=0.8)
ax5.set_title("Top 10 Countries by CAGR (2010–2021)")
ax5.set_xlabel("CAGR %")

# Chart 6 — Forecast
ax6 = axes[1,2]
forecast_years = pd.DataFrame(range(2022,2025), columns=["Year"])
ax6.plot(yearly["Year"], yearly["Total_Exports"],
         marker="o", color="#1B6B5C", label="Actual", linewidth=2)
ax6.plot(forecast_years["Year"],
         model.predict(forecast_years),
         marker="o", color="#D85A30", linestyle="--",
         label="Forecast", linewidth=2)
ax6.axvline(2021, color="gray", linestyle=":", linewidth=1)
ax6.set_title("Export Trend & Forecast")
ax6.set_ylabel("Export Value (USD Mn)")
ax6.legend()
ax6.yaxis.set_major_formatter(plt.FuncFormatter(lambda x,_: f"{x:,.0f}"))

plt.tight_layout()
chart_out = os.path.join(folder_path, "india_exports_charts.png")
plt.savefig(chart_out, dpi=150, bbox_inches="tight")
plt.show()

print(f"\n✓ Charts saved: {chart_out}")
print("\n✅ ALL DONE!")
print(f"   → india_exports_clean.xlsx")
print(f"   → india_exports_charts.png")
print("\nNext step: Upload india_exports_clean.xlsx to Tableau Public!")
```
